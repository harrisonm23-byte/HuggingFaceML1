```python
# Colab's tool for reading values saved under Secrets (the key icon)
from google.colab import userdata
# Hugging Face's login helper
from huggingface_hub import login

# Log in with the token saved as HF_TOKEN, so we can download gated models like Gemma
login(token=userdata.get('HF_TOKEN'))
```

```python
# PyTorch, the library that actually runs the model's math
import torch
# The "pipeline" bundles the model and its tokenizer into one easy-to-call object
from transformers import pipeline

# Download Gemma 3 1B (instruction-tuned) and load it onto the GPU
generator = pipeline(
    "text-generation",           # the task: continue a conversation with new text
    model="google/gemma-3-1b-it",  # which model to download from Hugging Face
    device="cuda",               # run on the GPU
    dtype=torch.float32,         # full precision: safe on a T4 and the 1B model easily fits
)
```

```python
# A conversation is a list of messages; this one has a single user question
messages = [{"role": "user", "content": "What is the capital of France?"}]

# Ask the model to reply, allowing at most 50 new tokens (word pieces)
result = generator(messages, max_new_tokens=50)

# The output is the whole conversation; the last message is the model's reply
print(result[0]["generated_text"][-1]["content"])
```
